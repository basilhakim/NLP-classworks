```python
# Step 1: Set Up the Environment
# Install necessary libraries
!pip install datasets
!pip install seaborn

# Download GloVe embeddings
!wget http://nlp.stanford.edu/data/glove.6B.zip

# Unzip the downloaded file
import zipfile
with zipfile.ZipFile('glove.6B.zip', 'r') as zip_ref:
    zip_ref.extractall()

# Step 2: Load the GloVe Embeddings
import numpy as np

def load_glove_embeddings(file_path):
    embeddings_index = {}
    with open(file_path, encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            coefs = np.asarray(values[1:], dtype='float32')
            embeddings_index[word] = coefs
    print(f'Loaded {len(embeddings_index)} word vectors.')
    return embeddings_index

# Specify the path to your GloVe file (e.g., 'glove.6B.100d.txt')
glove_file_path = 'glove.6B.100d.txt'  # Change to desired dimension (e.g., 50d, 100d, 200d, 300d)
embeddings_index = load_glove_embeddings(glove_file_path)

# Step 3: Find Nearest Words
from sklearn.metrics.pairwise import cosine_similarity

def find_nearest_words(word, embeddings_index, top_n=5):
    embedding_vector = embeddings_index.get(word)

    if embedding_vector is None:
        print(f'Word "{word}" not found in GloVe embeddings.')
        return []

    # Calculate cosine similarities with all other words
    similarities = {}
    for other_word, other_vector in embeddings_index.items():
        if other_word != word:  # Avoid comparing the word with itself
            similarity = cosine_similarity([embedding_vector], [other_vector])[0][0]
            similarities[other_word] = similarity

    # Get the top N most similar words
    nearest_words = sorted(similarities.items(), key=lambda item: item[1], reverse=True)[:top_n]

    return nearest_words

# Example usage
word_to_check = 'example'  # Replace with any word you want to check
nearest_words = find_nearest_words(word_to_check, embeddings_index)

print(f'Nearest words to "{word_to_check}":')
for similar_word, similarity in nearest_words:
    print(f'{similar_word}: {similarity:.4f}')
```
