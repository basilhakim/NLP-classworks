```python
import random
from gensim.models import Word2Vec
```

```python
# Create 100 random sentences
# Sample quotes from "Pride and Prejudice"
corpus = [
    "It is a truth universally acknowledged, that a single man in possession of a good fortune, must be in want of a wife.",
    "However little known the feelings or views of such a man may be on his first entering a neighbourhood, this truth is so well fixed in the minds of the surrounding families, that he is considered as the rightful property of some one or other of their daughters.",
    "My dear Mr. Bennet, said his wife to him one day, have you heard that Netherfield Park is let at last?",
    "Mr. Bingley was to dine with them that evening.",
    "Is he married or single?",
    "Oh! Single, my dear, to be sure! A single man of large fortune; four or five thousand a year.",
    "What a fine thing for our girls!",
    "How so? How can it affect them?",
    "My dear Mr. Bennet, replied his wife, how can you be so tiresome!",
    "You must know that I am thinking of his marrying one of them."
]
```

```python
corpus
```

```python
# Tokenize the sentences
sentences = [sentence.split() for sentence in corpus]
```

```python
sentences
```

```python

# Train CBOW model
model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, sg=0)
```

```python
# Example usage to find similar words
similar_words = model.wv.most_similar("man", topn=5)
print("Words similar to 'man':")
for word, similarity in similar_words:
    print(f"{word}: {similarity:.4f}")
```

```python
# Extract and display word embeddings for each word in the vocabulary
print("Word Embeddings:")
for word in model.wv.key_to_index:  # Iterate through all words in the vocabulary
    embedding = model.wv[word]  # Get the embedding for the word
    print(f"{word}: {embedding.tolist()}")  # Print the word and its embedding
```
